```python
import torch
import cv2
# from test_color import show_image
from PIL import Image
from matplotlib import pyplot as plt
from torchvision import transforms
import numpy as np

from torchsummary import summary
```

```python
img = cv2.imread(r'../cat.png')
```

```python
cv2.imshow("cat", img)
cv2.waitKey(0)
```

```python
img_lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB).astype(np.float32)
img_lab /= 255.0
cv2.imshow("lab", img_lab)
cv2.waitKey(0)
```

```python
lab_loaded = np.load("../cat_lab.npy")
cv2.imshow("cat_lab", lab_loaded)
cv2.waitKey(0)
```

```python
rgb = np.array([[123, 232, 80]], dtype=np.float32)
rgb = rgb.reshape((1, 1, 3))
rgb /= 255.0

rgb_convert = cv2.cvtColor(rgb, cv2.COLOR_RGB2LAB)
rgb_convert_reverse = cv2.cvtColor(rgb_convert, cv2.COLOR_LAB2BGR) * 255.0

print(rgb_convert)
print(rgb_convert_reverse)
```
